```python
import matplotlib.pyplot as plt
import numpy as np
```

```python
N = [10, 15, 20, 30]
aa = [.01, .05, .1, .2, .4]


for n in N:
    ekns = []
    for k in range(1, 11):
        ekn = .1 * k * (n - k)
        ekns.append(ekn)
    plt.plot(ekns, label=str(n))

plt.legend()
```
