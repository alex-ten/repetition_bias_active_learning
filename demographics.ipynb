```python
import pandas as pd

def display_(df, rows=None, cols=None, colwidth=10):
    with pd.option_context('display.max_rows', rows, 'display.max_columns', cols, 'display.max_colwidth', colwidth):  # more options can be specified also
        display(df)
```

```python
def extract_ids(file, save_to):
    df = pd.read_json(file, lines=True)
    ids = df.prolificPID.unique()
    with open(save_to, 'w+') as file:
        file.write(', '.join(ids))


def extract_ids2(file, save_to):
    df = pd.read_csv(file)
    ids = df['Participant id'].unique()
    with open(save_to, 'w+') as file:
        file.write(', '.join(ids))

extract_ids2('data/Mar2023/interleaved_demo.csv', 'data/Mar2023/interleaved_ids.txt')
```
