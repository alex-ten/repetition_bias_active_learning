# Compile id list

```python
groups = ['free', 'blocked', 'interleaved']
for group in groups:
    ids = []
    groups_to_exclude = groups[:]
    groups_to_exclude.pop(groups.index(group))
    for group_to_exclude in groups_to_exclude:
        df = pd.read_csv(f'pilots/pilots_Jan2023/prolific_{group_to_exclude}.csv')
        ids += df.loc[:, 'Participant id'].to_list()

    with open(f'pilots/pilots_Jan2023/blocklist_{group}.txt', 'w') as file:
        n = file.write(' '.join(ids))
```

```python
paths = [
    'pilots/pilots_Jan2023/prolific_free.csv',
    'pilots/pilots_Jan2023/prolific_blocked.csv',
    'pilots/pilots_Jan2023/prolific_interleaved.csv'
]
ids = []
for path in paths:
    df = pd.read_csv(path)
    ids += df.loc[df['Completion code'].eq('C15BMR3K'), 'Participant id'].to_list()

with open(f'pilots/pilots_Jan2023/all_ids.txt', 'w') as file:
    n = file.write(' '.join(ids))
```
